## Creating Dictionaries!

- The `{}` syntax creates an empty dictionary.
- You can also use `dict()`, but that seems rare in real life.
- Assignment and lookup use the same syntax as lists.

```python
phonebook = {}
phonebook['Empire'] = '588-2300'  # Empire is the key, 588-2300 is the value
phonebook['Emergency'] = '911'
phonebook['Jenny'] = '867-5309'
phonebook
```

```python
phonebook['Jenny']
```

```python
phonebook['Mattox']
```

```python
'Jenny' in phonebook
```

```python
'Mattox' in phonebook
```

```python
phonebook.keys()
```

```python
phonebook.values()
```

```python
for k in phonebook.keys():
    print(k,phonebook[k])
```
