## Higher Order Functions

- A function lives in a variable just like anything else!

```python
def inc(x):
    "Increment the input."
    return x + 1
```

```python
inc(10)
```

```python
inc
```

```python
abc = inc
```

```python
abc
```

```python
abc(10)
```

```python
def twice(f,x):
    return f(f(x))
```

```python
twice(abc,10)
```
