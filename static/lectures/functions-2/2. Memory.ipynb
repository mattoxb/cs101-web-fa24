## Memory
- If you **assign to a variable** the it is **local**.
- If you **only reference a variable** then it is **global**.

```python
x = 10
```

```python
def foo(a):
    y = a + x
    return y
```

```python
foo(20)
```

```python
def bar(a):
    x = a + 10
    return x
```

```python
bar(10)
```

```python
x
```

## Overriding Locals!

- The `global` keyword will override this behavior.

```python
def baz(a):
    global x
    x = a + 10
    return x
```

```python
baz(10)
```

```python
x
```

```python
timestamp = 0
```

```python
def tick(delta = 1):
    global timestamp
    timestamp = timestamp + delta
    return timestamp
```

```python
tick()
```

```python
tick()
```

```python
tick()
```

## It can even create variables!

- Don't do this unless you have a really really good reason for it.

```python
def goop(a):
    global z
    z = a
```

```python
goop(10)
```

```python
z
```
