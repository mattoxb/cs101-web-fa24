## Keyword Parameters

- Useful when there are usual values or lots of options that don't change much.
- In this example, `name` is a positional parameter, `greeting` and `excited` are keyword parameters.

```python
def greet(name, greeting='Hello,', excited=False):
    "Greet the user.  You can set the greeting and set whether or not we are greeting excitedly."
    out = f"{greeting} {name}"
    if excited:
        out = out + "!"
    else:
        out = out + "."
    return out
```

```python
greet("Jean-luc")
```

```python
greet("Jean-luc",excited=True)
```

```python
greet("Jean-luc", greeting="Bonjour,")
```

```python
## You can get away with not specifying the names if you keep the order right.

greet("Jean-luc", "Bonjour,", True)
```

```python
## You cannot get away with specifying a keyword before the positional arguments.
greet(greeting="Bonjour,", "Jean-luc")
```
