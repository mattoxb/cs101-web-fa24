## What happens in memory?

```python
x = 10
```

```python
def foo(a):
    y = a + x
    return y
```

```python
b = foo(20)
b
```

In Memory:

| Scope | Name | Value |
| ----- | ---- | ----- |
| Global | x    | 10  |
|        | b    | None |

Then `foo` gets called.

| Scope | Name | Value |
| ----- | ---- | ----- |
| Global | x    | 10  |
|        | b    | None |
|-------|------|-------|
| `foo` | a    | 10    |
|       | y    | 10 + 10 = 20 |
|       | (result) | 20 |

Then `foo` returns, putthing the return result into `b`.

| Scope | Name | Value |
| ----- | ---- | ----- |
| Global | x    | 10  |
|        | b    | 20 |


```python
def bar(a):
    x = a + 10
    return x
```

```python
c = bar(b)
c
```

In Memory:

| Scope | Name | Value |
| ----- | ---- | ----- |
| Global | x    | 10  |
|        | b    | 20 |
|        | c    | None |

Then `bar` gets called.

| Scope | Name | Value |
| ----- | ---- | ----- |
| Global | x    | 10  |
|        | b    | 20 |
|        | c    | None |
|-------|------|-------|
| `bar` | a    | 20    |
|       | x    | 20 + 10 = 30 |
|       | (result) | 30 |

Then `bar` returns.

| Scope | Name | Value |
| ----- | ---- | ----- |
| Global | x    | 10  |
|        | b    | 20 |
|        | c    | 30 |


Did you notice there are two variables named `x` this time?



## The Rule:

- If a function **only reads** from a variable it will use the **global** version.
- If a function **ever tries to write** to a variable it will always treat it as a **local** version.
- No trying to have it both ways!!

```python
def baz(a):
    "This is a broken function to demonstrate the difference between local and global variables."
    b = a + x
    x = x + 1
    b = b + x
    return b
```

```python
baz(10)
```
