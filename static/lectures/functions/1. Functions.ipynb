## Functions

- If you are repeating a computation, you will want to write it in a function.
- Syntax:
```
def fname(p1,p2):
    "This is an example function."
    body
    return result
```
- The `def` keyword warns Python you are creating a function.
- The `p1` and `p2` names are called *parameters*.
- The string on the next line is called the "doc string".  (Short for *documentation string*.)
- The `return` keyword declares what the final result should be.

````python
def plus1(a):
    "Adds 1 to parameter a."
    a = a + 1
    return a
````

````python
plus1(10)  # The 10 is called an "argument"
````

````python
plus1(20 + 49)
````

````python
# Note that `a` does not exist in the main program: only in the function!
a
````

````python
print(plus1.__doc__)
````

````python
def add(a,b):
    "Returns the sum of a and b."
    return a + b
````

````python
add(10,20)
````

````python
add(add(10,20),plus1(40))
````

- `add(10,20)` runs first and makes the first argument 30.
  - Result: `add(30,plus1(40))`
- `plus1(40)` runs next and makes the second argument 41.
  - Result: `add(30,41)`
- `add(30,41)` runs last.
  - Result: 71

## Example: can you make this code shorter?

- Someone needed to find the hypotneuses of 4 triangles.  So they wrote this code.
- They used cut-and-paste-and-edit, and actually messed a few things up.
- You know three ways to will really shorten this code!

````python
x1 = 10
x2 = 20
x3 = 30
x4 = 40
````

````python
y1 = 40
y2 = 44
y3 = 20
y4 = 30
````

````python
# Not only is this code longer than it should be, it's also wrong.
print(f"Hypotnus is {(x1*x1 + y1*y1)**0.5:.4f}.")
print(f"Hypotnus is {(x2*x1 + y2*y3)**0.5:.4f}.")
print(f"Hypotnus is {(x3*x3 + y2*y3)**0.5:.4f}.")
print(f"Hypotnus is {(x4*x4 + y4*y4)**0.5:.4f}.")
````
