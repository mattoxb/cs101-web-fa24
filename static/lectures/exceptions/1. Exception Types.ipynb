## Assertions and Stack Traces
- Run the code below 

```python
def foo():
    assert 1 == 2
    
def bar():
    foo()
    
def baz(x):
    if x == 1:
        bar()
    else:
        baz()
```

```python
baz(1)
```

```python
baz(0)
```
