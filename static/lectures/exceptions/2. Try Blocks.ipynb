## Try Blocks

You can tell Python to "try" to do something.  If it fails, it you can tell it how to recover.

```python
mylist = [10,20,50]
```

```python
def safeLookup(i):
    try:
        i = int(input("Give me an index: "))
        return mylist[i]
    except:
        print("Ug.  Something went wrong.")
```

```python
safeLookup()
```

```python
safeLookup()
```

## Getting Specific

Remember those error types before?  We can catch them by name!

```python
def safeLookup():
    try:
        i = int(input("Give me an index: "))
        return mylist[i]
    except IndexError:
        print("It needs to be from 0 to 3")
    except ValueError:
        print("Please stick with integers.")
    except Exception as e:
        print("Ug.  Something went wrong.")
        print(e)
```

```python
safeLookup()
```
