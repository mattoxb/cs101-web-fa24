## Iterators!

There are quite a few Python functions that create iterators for us.  Today we will talk about the most common one, and one extra one.
- `range` creates a list.  There are three forms!
   - `range(n)` creates an iterator from 0 to n-1.
   - `range(a,b)` creates an iterator from a to b-1
   - `range(a,b,n)` creates an iterator from a to b-1, but skips ahead by n instead of one.
- `zip` pairwise merges two lists together 

## f-Strings!
- Python let's you create string with formatting in several ways.  The f-string is the one we will use most:

```python
print(f"This is the answer: {10 + 32}")
```

```python
print(f"This is the answer: {10 + 30:04d}.") # 04 = 4 digits with leading zeros, d = integers (d stands for decimal)
```

```python
print(f"This is the answer: {10 + 30: 4d}.") #  4 = 4 digits with leading spaces, d = integers (d stands for decimal)
```

```python
y = [8,6,7,5,3,0,9]
```

```python
for number in y:
    print(f"The number is {number}")
```

```python
y = range(10)
```

## The Zip function

```python
# Code 3
for a,b in [[1,10],[2,20],[3,30],[4,40]]:
    print(a,b)
```

```python
# Code 4a
pairs = zip(range(4),range(0,40,10))
pairs
```

```python
for a,b in pairs:
    print(f"a is {a} and b is {b}.")
```
