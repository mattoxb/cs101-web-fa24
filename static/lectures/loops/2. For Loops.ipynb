## How they work

- Keyword `for`, then a variable name (called the *loop variable*), the keyword `in`, and then a collection.
- The **loop body** is indented, just like with `if` statements.
- Inside a loop, we have two special keywords:
  - `break` will stop the loop completely and move to what's after the loop.
  - `continue` will skip the current element and go to the next one.

```python
# Code 6
a = 4
for i in [2,3,4,5,6,7,8]:
    if i == a:
        break
    print(i)
print("Done")
```

```python
# Code 6
a = 4
for i in [2,3,4,5,6,7,8]:
    if i == a:
        continue
    print(i)
print("Done")
```
