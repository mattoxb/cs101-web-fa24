## Lists!

- You can collect a bunch of things together by using a `list`.
- You can access the elemets of a list the same way you do with strings!
- *Unlike strings* you can also assign to individual elements

```python
xx = [2,4,6,5,7,9]
```

```python
xx[0]
```

```python
xx[0:3]
```

```python
xx[0] = 999
```

```python
xx
```

```python
xx[-1]
```

```python
xx[-3:-1]
```

```python
xx[-3:]
```

```python
xx + xx
```

## List Operations
- You can take the length, add, and sort

```python
len(xx)
```

```python
xx.append(99)
```

```python
xx
```

```python
## The `sort` method does not return the new list: it updates the original.
xx.sort()
```

```python
xx
```

```python
xx + [3,3,3]
```

## A surprise!
- Assigning a list variable to another list variable does **not** create a new list!

```python
yy = xx
```

```python
xx
```

```python
yy
```

```python
xx[0] = 123
```

```python
yy
```

```python
# If you really need a copy, use copy!
```

```python
zz = yy.copy()
```

```python
yy
```
