### Function 2

```python
def max(xx):
    m = 0
    for x in xx:
        if x < m:
            m = x
        return m
```

### Function 3

```python
import random
def estPi(runs = 1000):
    p_in = 0
    p_all = 0
    for _ in range(runs):
        x = random.random()
        y = random.random()
        if x + y < 1:
            p_in += 1
        p_all += 1
        
        return p_in / p_all
```
