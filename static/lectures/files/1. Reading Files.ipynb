## Reading Files

- Let's create a file and then try to read in the data.
- The `open(fname)` command will open a file that is in the same directory to enable you to read it.
- It will return a /file object/, also known as a /file handle/

```python
f = open('fear.txt')
```

```python
# There are three bits of information here.  What are they?
f
```

```python
f.read()
```

```python
f.read()
```

```python
f.close() # This can be important to do!
```

```python
f = open('fear.txt')
```

```python
f.readlines()
```

```python
# What happens if you open a non-existent file?
f = open("404.txt")
```

```python
with open('fear.txt') as file:
    num = 1
    for row in file.readlines():
        print(f"{num}: {row}")  # Do you remember format strings?
        num += 1
```
